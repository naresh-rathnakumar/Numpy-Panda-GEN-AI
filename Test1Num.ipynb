```python
a = int(input("Enter salary 1"))
b = int(input("Enter salary 2"))
c = int(input("Enter salary 3"))
d = int(input("Enter salary 4"))
e = int(input("Enter salary 5"))
f = int(input("Enter salary 6"))
salaries=np.array([a,b,c,d,e,f])
uniquesalaries = np.unique(salaries)
sortedsalaries = np.sort(uniquesalaries)
highest = sortedsalaries[-1]
secondhighest = sortedsalaries[-2]
print("Highest Salary:", highest)
print("Second Highest Salary:", secondhighest)
```

```python
quantity = np.array([[2, 3],[4, 5]])
price = np.array([[100, 200],[150, 250]])
print("Quantity Matrix:",quantity)
print("Price Matrix:",price)
print("Quantity Transpose:",quantity.T)
print("Matrix Multiplication:",quantity@price)
```

```python
marks = np.array([40, 50, 60, 70, 80, 90, 100])
print("Original Marks:",marks)
normalized = (marks - np.min(marks)) / (np.max(marks)-np.min(marks))
print("Normalized Marks:",np.round(normalized,4))
```

```python
sales = np.array([[10, 12, 15, 11, 18],[20, 18, 25, 22, 30],[15, 17, 14, 20, 19],[8, 10, 12, 9, 15]])
products = ["Laptop", "Mobile", "Tablet", "Monitor"]
days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]
print("Product-wise Sales:",np.sum(sales, axis=1))
print("Day-wise Sales:",np.sum(sales, axis=0))
bestproduct = np.argmax(np.sum(sales, axis=1))
print("Best Selling Product:", products[bestproduct])
print("Total Sales:", np.sum(sales, axis=1)[bestproduct])
bestday = np.argmax(np.sum(sales, axis=0))
print("Best Sales Day:", days[bestday])
print("Total Sales:", np.sum(sales, axis=0)[bestday])
print("Highest Single-Day Sale:", np.max(sales))
print("Sales above 20:", sales[sales > 20])
print("Overall Average Sales:", np.mean(sales))
producttotal = np.sum(sales, axis=1)
print("Products:")
print(np.array(products)[producttotal > 80])
```
